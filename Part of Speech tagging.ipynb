# Part of speech tagging
* POS tagging is the process of assiging a part of speech (noun, verb, adj etc) to each word in  
a sentence, based on its context and defination.

```python
sentence = "The Eiffel Tower is a wrought-iron lattice tower on the Champ de Mars in Paris, France. It is named after the engineer Gustave Eiffel, whose company designed and built the tower from 1887 to 1889."
```

```python
import nltk
words=nltk.word_tokenize(sentence)
```

```python
tag_elements=nltk.pos_tag(words)
```

```python
nltk.ne_chunk(tag_elements)  #download "maxent_ne_chunker" & "word"
```

```python
nltk.download('maxent_ne_chunker')
```

```python
nltk.ne_chunk(tag_elements).draw()
```
