```python
corpus="""Hello Welcome, to Prashant Bhosale's NLP Study.
Please do watch the entire course! to become expert in NLP"""
```

```python
print(corpus)
```

```python
## Tokenization
## Sentence --> paragraphs 
from nltk.tokenize import sent_tokenize
```

```python
documents=sent_tokenize(corpus)
```

```python
print(documents)
```

```python
#required for above task
import nltk
nltk.download('punkt') 
```

```python
type(documents)
```

```python
for sentence in documents:
    print(sentence)
```

```python
# tokenization
# Paragraph --> words
# Sentenece --> words

from nltk.tokenize import word_tokenize
```

```python
word_tokenize(corpus)
```

```python
for sentence in documents:
    print(word_tokenize(sentence))
```

```python
from nltk.tokenize import wordpunct_tokenize
#wordpunct_tokenize : seperate out punctuations like 's, consider this to seperate word.
```

```python
wordpunct_tokenize(corpus) 
```
