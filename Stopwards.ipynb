🔹 What are Stopwords?  
* Stopwords are the most common words in a language that usually carry little meaningful
information when analyzing text.  
* for example : a, an, the, is, are, was, were, in, on, at, by, for, with, to, from, and, or, of, this, that, it, etc.
* Why remove stopwords?Stopwords are removed in many NLP tasks to:
    - Reduce noise
    - Speed up processing
    - Focus on keywords and meaningful content
* When not to remove stopwords?
    - In sentiment analysis, words like not, never, isn’t are crucial.
    - In translation, summarization, stopwords may be important for context.
    - ✅ Remove for keyword extraction or topic modeling
    - ❌ Keep for sentiment/context-aware tasks
* Real-World Use Cases  
✅ Text Classification (reduce feature space)  
✅ Information Retrieval (like search engines)  
✅ Topic Modeling  
✅ WordClouds  
✅ Plagiarism Detection

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.stem import WordNetLemmatizer

# Download resources (run only once)
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')

lemmatizer = WordNetLemmatizer()

# Input text
text = """
Today, I stand before you to speak about one of India’s most inspiring personalities – Dr. APJ Abdul Kalam.
His journey from the small town of Rameswaram to becoming India’s President is a testament to the power of dreams and determination.

Born into a modest family, young Kalam had to sell newspapers to support his family’s income. However, his passion for learning never dimmed.
He studied physics and aerospace engineering, which laid the foundation for his remarkable scientific career.

At ISRO and DRDO, Kalam became the architect of India’s missile program. His leadership in developing the SLV-III and various missile systems earned him the title “Missile Man of India.”
But what set him apart wasn’t just his scientific achievements; it was his vision for India’s future.

As President, he connected with people, especially the youth, in a way few leaders have.
He didn’t just occupy Rashtrapati Bhavan; he transformed it into a centre of inspiration. His dream of a developed India, captured in his vision of “India 2020,” inspired millions.
"""

# Sentence tokenization
sentences = nltk.sent_tokenize(text)

# Lemmatization & Stopword Removal
for i in range(len(sentences)):
    words = nltk.word_tokenize(sentences[i])
    words = [lemmatizer.lemmatize(word.lower()) 
             for word in words 
             if word.lower() not in set(stopwords.words('english')) and word.isalnum()]
    sentences[i] = ' '.join(words)

# Output cleaned sentences
for line in sentences:
    print(line)
```
