## PorterStemmer
Porter Stemmer is a rule-based (word reduction based on set of rules) stemming algorithm used to reduce English words to their root forms.   It's widely used in NLP for text normalization and works well when performance matters more than linguistic accuracy.

Simple and fast	--> Stems may not be valid words  
Reduces vocabulary size -->	May over-stem (lose meaning)

Porter stemmer algorithm prioritizes rule-based reduction, not correctness of resulting stems.

```python
from nltk.stem import PorterStemmer
```

```python
stemming=PorterStemmer()
```

```python
for word in words:
    print(word+"--->"+stemming.stem(word))
```

```python
# some of above word not have correct meaning that is measure disadvantage of stemming.
# like histori & easili. 
```

```python
stemming.stem("Congratulations")
```

```python
stemming.stem("sitting") #seat expected
```

## RegexpStemmer 
* RegexpStemmer (Regular Expression Stemmer) is a customizable stemming class in the NLTK library that allows you to define your **own stemming rules** using regular expressions.
* Unlike PorterStemmer or SnowballStemmer (which follow fixed algorithms), RegexpStemmer gives you full control over how words are stemmed by applying pattern-based rules you specify.
* morphological appix : regular expression

```python
from nltk.stem import RegexpStemmer
```

```python
reg_stemmer=RegexpStemmer('ing$|s$|e$|able$', min=4) 
#regular expression madhe we write our own rules
```

```python
reg_stemmer.stem('eating')
```

```python
reg_stemmer.stem('ingeating') #('ing|s$|e$|able$', min=4) it will remove both ing
```

## Snowball Stemmer
* The Snowball Stemmer is an improved version of the Porter Stemmer, created by the same person — Martin Porter — to be more accurate, readable, and consistent across multiple languages.
* Handles more morphological variations. Supports multiple languages (unlike Porter which only supports English)
* 

```python
from nltk.stem import SnowballStemmer

SbStemmer=SnowballStemmer('english')
```

```python
for word in words:
    print(word+"--->"+SbStemmer.stem(word))

#performance better than previous
# writing --> write
```

```python
stemming.stem("fairly"),stemming.stem("sportingly")
```

```python
SbStemmer.stem("fairly"),SbStemmer.stem("sportingly") #exact word (meaningful) we get
```
