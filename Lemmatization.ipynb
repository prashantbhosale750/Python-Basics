## Lemmaitzation
* Lemmatization is the process of reducing a word to its base or dictionary form, called a lemma, using vocabulary and morphological analysis.
* Unlike stemming (which simply chops off suffixes), lemmatization returns valid words that you would find in a dictionary.
* Use lemmatization when:
  1. grammatical correctness
  2. you are doing text classification, info retrival
  3. chatbots, search engines
* 

```python
import nltk
nltk.download('wordnet')
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()
```

```python

words=["eating", "eat","eaten", "writing", "wites", "programming", "programs","histori","finally","finalized","easily"] 
```

```python
words=["eating", "eat","eaten","running" ,"writing", "wites", "programming", "programs","histori","finally","finalized","easily"] 

for word in words: 
    print(f"{word} → {lemmatizer.lemmatize(word,pos='v')}")
```

```python
#nouns
lemmatizer.lemmatize("cars")         
lemmatizer.lemmatize("children")     
lemmatizer.lemmatize("geese")       
```

```python
#verbs
lemmatizer.lemmatize("running", pos="v")  # → "run"
lemmatizer.lemmatize("studies", pos="v")   #→ "study"
lemmatizer.lemmatize("ate", pos="v")       #→ "eat"
lemmatizer.lemmatize("written", pos="v")   #→ "write"
```

```python
#adjectives
lemmatizer.lemmatize("better", pos="a")   # → "good"
lemmatizer.lemmatize("best", pos="a")     # → "good"
lemmatizer.lemmatize("happier", pos="a")  # → "happy"
```

```python
#default behaviour

lemmatizer.lemmatize("running")    # → "running"   # Without POS, assumes noun
lemmatizer.lemmatize("flies")      # → "fly"       # OK, since "flies" as noun
lemmatizer.lemmatize("better")      # → "better"    # Needs POS to convert
```
