```python
#Que. 1

def str_manipulation(s):
  s=s.replace('a','b')
  s=s.replace('b','c')
  s=s.replace('c','a')
  return s
s='abcabc'
print(str_manipulation(s)) 
#1. abcabc, 2.bccbcc, 3.aaaabb, 4. aaaaaa
```

```python
#Que.2 : Pascal Triangle 
#using binomial coeffient : number = number * (i-j)//(j+1)

n = int(input("Enter the number: "))
for i in range(n):
    number = 1   
    for j in range(i+1):
        print(number, end=" ")
        number = number * (i-j)//(j+1)
    print()
```
